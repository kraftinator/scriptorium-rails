# Scriptorium — Convert safetensors → mlmodel

Kraken 7 saves models as `.safetensors`, but kraken 6 (the newest version that runs on Raspberry Pi 4's ARM CPU) can only load the older `.mlmodel` (Core ML) format. This notebook loads the trained safetensors on Kaggle (where kraken 7 works) and re-serializes it as mlmodel for use on the Pi.

Just **Save Version → Save & Run All (Commit)** — CPU-only, ~30 seconds. The output `.mlmodel` lands in the notebook's Output tab.

```python
# Pull the safetensors model from the git-LFS-tracked file in the repo.
%cd /kaggle/working
!rm -rf scriptorium-rails
!GIT_LFS_SKIP_SMUDGE=1 git clone https://github.com/kraftinator/scriptorium-rails.git
%cd scriptorium-rails
!git lfs pull -I data/models/barton_htr.safetensors
!ls -lh data/models/
```

```python
# Load safetensors with kraken 7's loader, then write as mlmodel with the writer.
from kraken.models.loaders import load_safetensors
from kraken.models.writers import write_coreml

src = 'data/models/barton_htr.safetensors'
dst = '/kaggle/working/barton_htr.mlmodel'

models = load_safetensors(src)
print(f'loaded {len(models)} model(s): {[type(m).__name__ for m in models]}')

# write_coreml signature is (obj: list[TorchVGSLModel], path). Pass the list.
write_coreml(models, dst)
print(f'wrote {dst}')
!ls -lh /kaggle/working/barton_htr.mlmodel
```
